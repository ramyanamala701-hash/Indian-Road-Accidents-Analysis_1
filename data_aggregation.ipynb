# Data Grouping and Aggregation


## Import Required Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the Dataset


```python
df = pd.read_csv("../data/india_road_accidents_dataset.csv")
print("Dataset loaded successfully")
```

## Create Derived Casualty Features


```python
df['total_injuries'] = df['severe_injuries'] + df['minor_injuries']
df['total_casualties'] = df['fatalities'] + df['total_injuries']
```

## Accidents by State/UT


```python
state_accidents = (
    df.groupby('state_ut')
      .size()
      .reset_index(name='accident_count')
      .sort_values('accident_count', ascending=False)
)
state_accidents
```

## Accidents by Severity


```python
severity_summary = (
    df.groupby('severity')
      .size()
      .reset_index(name='accident_count')
      .sort_values('accident_count', ascending=False)
)
severity_summary
```

## Accidents by Weather


```python
weather_summary = (
    df.groupby('weather')
      .size()
      .reset_index(name='accident_count')
      .sort_values('accident_count', ascending=False)
)
weather_summary
```

## Accidents by Vehicle Type


```python
vehicle_summary = (
    df.groupby('vehicle_type')
      .size()
      .reset_index(name='accident_count')
      .sort_values('accident_count', ascending=False)
)
vehicle_summary
```

## Fatalities by State/UT


```python
fatalities_by_state = (
    df.groupby('state_ut')['fatalities']
      .sum()
      .reset_index()
      .sort_values('fatalities', ascending=False)
)
fatalities_by_state
```

## Average Speed by Road Type


```python
speed_by_road = (
    df.groupby('road_type')['estimated_speed_kmph']
      .mean()
      .reset_index()
      .sort_values('estimated_speed_kmph', ascending=False)
)
speed_by_road
```

## Average Casualties by Severity


```python
casualties_by_severity = (
    df.groupby('severity')['total_casualties']
      .mean()
      .reset_index()
      .sort_values('total_casualties', ascending=False)
)
casualties_by_severity
```

## Overall Aggregated Summary


```python
overall_summary = df.agg({
    'fatalities': ['sum', 'mean', 'max'],
    'total_injuries': ['sum', 'mean', 'max'],
    'total_casualties': ['sum', 'mean', 'max'],
    'estimated_speed_kmph': ['mean', 'max']
})
overall_summary
```
