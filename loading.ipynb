# Data Loading and Inspection

This notebook loads the Indian road accidents dataset and performs initial inspection of its structure, columns, data types, missing values, duplicate records, basic statistics, and unique values.

```python
import pandas as pd
```

## Load the Dataset

```python
# Load the dataset
df = pd.read_csv("../data/india_road_accidents_dataset.csv")
print("Dataset loaded successfully")
```

## View the First 5 Records

```python
# Display first 5 records
df.head()
```

## Check Dataset Shape

```python
# Display number of rows and columns
print("Dataset Shape:", df.shape)
```

## Check Column Names

```python
# Display column names
print("Column Names:")
print(df.columns.tolist())
```

## Inspect Dataset Information

```python
# Display data types and non-null information
df.info()
```

## Check Missing Values

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

## Check Duplicate Records

```python
# Check the number of duplicate rows
print("Duplicate Records:", df.duplicated().sum())
```

## Statistical Summary

```python
# Display basic statistical information for numerical columns
df.describe()
```

## Check Unique Values

```python
# Display the number of unique values in each column
print("Unique Values:")
print(df.nunique())
```
