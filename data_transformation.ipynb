# Data Transformation and Feature Engineering


## Import Required Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the Dataset


```python
df = pd.read_csv("../data/india_road_accidents_dataset.csv")
print("Dataset loaded successfully")
```

## Convert Date and Time Column


```python
df['date_time'] = pd.to_datetime(df['date_time'])
print(df['date_time'].head())
```

## Create Speed Difference Feature


```python
df['speed_difference_kmph'] = (
    df['estimated_speed_kmph'] - df['speed_limit_kmph']
)
df[['speed_limit_kmph', 'estimated_speed_kmph', 'speed_difference_kmph']].head()
```

## Create Speeding Feature


```python
df['speeding'] = np.where(
    df['estimated_speed_kmph'] > df['speed_limit_kmph'],
    'Yes',
    'No'
)
df['speeding'].value_counts()
```

## Create Total Injuries Feature


```python
df['total_injuries'] = (
    df['severe_injuries'] + df['minor_injuries']
)
df[['severe_injuries', 'minor_injuries', 'total_injuries']].head()
```

## Create Total Casualties Feature


```python
df['total_casualties'] = (
    df['fatalities'] + df['total_injuries']
)
df[['fatalities', 'total_injuries', 'total_casualties']].head()
```

## Check Final Dataset Structure


```python
print("Final Dataset Shape:", df.shape)
print("New Columns:")
print(['speed_difference_kmph', 'speeding', 'total_injuries', 'total_casualties'])
```
