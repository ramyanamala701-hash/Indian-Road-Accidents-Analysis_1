# Data Visualization


## Import Required Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the Dataset


```python
df = pd.read_csv("../data/india_road_accidents_dataset.csv")
print("Dataset loaded successfully")
```

## Prepare Derived Features


```python
df['speed_difference_kmph'] = df['estimated_speed_kmph'] - df['speed_limit_kmph']
df['speeding'] = np.where(df['estimated_speed_kmph'] > df['speed_limit_kmph'], 'Yes', 'No')
df['total_injuries'] = df['severe_injuries'] + df['minor_injuries']
df['total_casualties'] = df['fatalities'] + df['total_injuries']
```

## 1. Accidents by Year


```python
year_counts = df.groupby('year').size()
plt.figure(figsize=(8, 5))
plt.plot(year_counts.index, year_counts.values, marker='o')
plt.title('Accidents by Year')
plt.xlabel('Year')
plt.ylabel('Number of Accidents')
plt.grid(True)
plt.show()
```

## 2. Top 10 States/UTs by Accident Count


```python
state_counts = df['state_ut'].value_counts().head(10).sort_values()
plt.figure(figsize=(10, 6))
plt.barh(state_counts.index, state_counts.values)
plt.title('Top 10 States/UTs by Accident Count')
plt.xlabel('Number of Accidents')
plt.ylabel('State/UT')
plt.show()
```

## 3. Accident Severity


```python
severity_counts = df['severity'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(severity_counts.index, severity_counts.values)
plt.title('Accident Severity Distribution')
plt.xlabel('Severity')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=30)
plt.show()
```

## 4. Top 10 Primary Causes


```python
cause_counts = df['primary_cause'].value_counts().head(10).sort_values()
plt.figure(figsize=(10, 6))
plt.barh(cause_counts.index, cause_counts.values)
plt.title('Top 10 Primary Causes of Recorded Accidents')
plt.xlabel('Number of Accident Records')
plt.ylabel('Primary Cause')
plt.show()
```

## 5. Vehicle Types


```python
vehicle_counts = df['vehicle_type'].value_counts()
plt.figure(figsize=(10, 6))
plt.bar(vehicle_counts.index, vehicle_counts.values)
plt.title('Accidents by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Number of Accident Records')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

## 6. Weather Conditions


```python
weather_counts = df['weather'].value_counts()
plt.figure(figsize=(9, 5))
plt.bar(weather_counts.index, weather_counts.values)
plt.title('Accidents by Weather Condition')
plt.xlabel('Weather Condition')
plt.ylabel('Number of Accident Records')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

## 7. Road Surface Conditions


```python
surface_counts = df['road_surface'].value_counts()
plt.figure(figsize=(9, 5))
plt.bar(surface_counts.index, surface_counts.values)
plt.title('Accidents by Road Surface Condition')
plt.xlabel('Road Surface')
plt.ylabel('Number of Accident Records')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

## 8. Lighting Conditions


```python
lighting_counts = df['lighting'].value_counts()
plt.figure(figsize=(9, 5))
plt.bar(lighting_counts.index, lighting_counts.values)
plt.title('Accidents by Lighting Condition')
plt.xlabel('Lighting Condition')
plt.ylabel('Number of Accident Records')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

## 9. Estimated Speed vs Speed Limit


```python
plt.figure(figsize=(8, 6))
plt.scatter(df['speed_limit_kmph'], df['estimated_speed_kmph'])
min_speed = min(df['speed_limit_kmph'].min(), df['estimated_speed_kmph'].min())
max_speed = max(df['speed_limit_kmph'].max(), df['estimated_speed_kmph'].max())
plt.plot([min_speed, max_speed], [min_speed, max_speed], linestyle='--', label='Speed Limit Reference')
plt.title('Estimated Speed vs Speed Limit')
plt.xlabel('Speed Limit (km/h)')
plt.ylabel('Estimated Speed (km/h)')
plt.legend()
plt.show()
```

## 10. Top 10 States/UTs by Fatalities


```python
fatality_counts = df.groupby('state_ut')['fatalities'].sum().sort_values(ascending=False).head(10).sort_values()
plt.figure(figsize=(10, 6))
plt.barh(fatality_counts.index, fatality_counts.values)
plt.title('Top 10 States/UTs by Fatalities')
plt.xlabel('Total Fatalities')
plt.ylabel('State/UT')
plt.show()
```
