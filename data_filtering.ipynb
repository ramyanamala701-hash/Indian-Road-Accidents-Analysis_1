# Data Filtering


## Import Required Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the Dataset


```python
df = pd.read_csv("../data/india_road_accidents_dataset.csv")
print("Dataset loaded successfully")
```

## Filter Fatal Accidents


```python
fatal_accidents = df[df['severity'] == 'Fatal']
print("Number of Fatal Accidents:", len(fatal_accidents))
fatal_accidents.head()
```

## Filter Accidents Above the Speed Limit


```python
high_speed_accidents = df[
    df['estimated_speed_kmph'] > df['speed_limit_kmph']
]
print("Accidents Above Speed Limit:", len(high_speed_accidents))
high_speed_accidents.head()
```

## Filter Alcohol-Involved Accidents


```python
alcohol_accidents = df[df['alcohol_involved'] == True]
print("Alcohol-Involved Accidents:", len(alcohol_accidents))
alcohol_accidents.head()
```
